```python
# Student Marks Analysis Using Python

# Project Overview
This project analyzes student marks using Python and basic statistical calculations.

# Objectives
- Store student names and their marks.
- Calculate total and average marks.
- Identify the highest and lowest marks.
- Display individual student performance.
- Assign performance categories.
- Visualize student marks using a bar chart.

# Tools Used
- Python
- Google Colab / Jupyter Notebook
- Matplotlib

# Expected Outcome
A Python-based analysis that summarizes student performance and presents the results in a clear, understandable format.
```

```python
# Import the library required for visualization
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Student names and their marks out of 100
students = ["Rahul", "Priya", "Arjun", "Sneha", "Vasanth"]
marks = [85, 92, 76, 88, 95]

# Display the dataset
print("STUDENT MARKS DATASET")
print("-" * 30)

for name, mark in zip(students, marks):
    print(f"{name}: {mark}/100")
```

```python
# Check whether the dataset is valid
if len(students) != len(marks):
    raise ValueError("Each student must have a corresponding mark.")

if len(marks) == 0:
    raise ValueError("The marks list cannot be empty.")

if not all(isinstance(mark, (int, float)) and not isinstance(mark, bool)
           and 0 <= mark <= 100 for mark in marks):
    raise ValueError("Every mark must be a number between 0 and 100.")

if len(set(students)) != len(students):
    raise ValueError("Student names must be unique.")

print("Data validation successful!")
```

```python
# Calculate the required statistics
total_marks = sum(marks)
average_marks = total_marks / len(marks)
highest_marks = max(marks)
lowest_marks = min(marks)
mark_range = highest_marks - lowest_marks

# Identify students with the highest and lowest marks
top_students = [
    students[i] for i, mark in enumerate(marks)
    if mark == highest_marks
]

lowest_students = [
    students[i] for i, mark in enumerate(marks)
    if mark == lowest_marks
]

# Display results
print("STUDENT MARKS ANALYSIS")
print("=" * 35)
print("Number of students:", len(students))
print("Total marks:", total_marks)
print("Average marks:", round(average_marks, 2))
print("Highest marks:", highest_marks)
print("Lowest marks:", lowest_marks)
print("Marks range:", mark_range)
print("Top performer(s):", ", ".join(top_students))
print("Lowest scorer(s):", ", ".join(lowest_students))
```

```python
# Categorize student performance
def get_performance(mark):
    if mark >= 90:
        return "Excellent"
    elif mark >= 80:
        return "Very Good"
    elif mark >= 70:
        return "Good"
    elif mark >= 50:
        return "Average"
    else:
        return "Needs Improvement"

print("INDIVIDUAL STUDENT PERFORMANCE")
print("=" * 45)

for name, mark in zip(students, marks):
    percentage = (mark / 100) * 100
    category = get_performance(mark)

    print(
        f"{name}: {mark}/100 | "
        f"Percentage: {percentage:.1f}% | "
        f"Performance: {category}"
    )
```

```python
# Create a structured table using pandas
import pandas as pd

results = pd.DataFrame({
    "Student Name": students,
    "Marks": marks,
    "Percentage": [round((mark / 100) * 100, 2) for mark in marks],
    "Performance": [get_performance(mark) for mark in marks]
})

display(results)
```

```python
# Bar chart showing individual student marks
plt.figure(figsize=(10, 5))

bars = plt.bar(students, marks)

plt.title("Student Marks Analysis", fontsize=16)
plt.xlabel("Student Name")
plt.ylabel("Marks out of 100")
plt.ylim(0, 110)
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Display marks above each bar
for bar, mark in zip(bars, marks):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        mark + 1,
        str(mark),
        ha="center"
    )

plt.tight_layout()
plt.show()
```

```python
# Compare each student's marks with the class average
print("COMPARISON WITH CLASS AVERAGE")
print("=" * 40)

for name, mark in zip(students, marks):
    difference = mark - average_marks

    if difference > 0:
        status = "Above average"
    elif difference < 0:
        status = "Below average"
    else:
        status = "Equal to average"

    print(
        f"{name}: {mark} marks | "
        f"Difference: {difference:+.2f} | {status}"
    )
```

```python
# Export the analysis results
csv_filename = "student_marks_results.csv"

results.to_csv(csv_filename, index=False)

print(f"Results saved to {csv_filename}")

# Download the CSV file to your computer
from google.colab import files
files.download(csv_filename)
```

```python
# Generate a final summary report
print("=" * 45)
print("FINAL PROJECT SUMMARY")
print("=" * 45)

print(f"Total students analyzed: {len(students)}")
print(f"Total marks: {total_marks}")
print(f"Class average: {average_marks:.2f}/100")
print(f"Highest marks: {highest_marks}/100")
print(f"Lowest marks: {lowest_marks}/100")
print(f"Top performer(s): {', '.join(top_students)}")
print(f"Lowest scorer(s): {', '.join(lowest_students)}")

print("\nConclusion:")
print(
    "The program successfully analyzed student marks, "
    "calculated key statistics, categorized performance, "
    "and generated a visualization."
)
```
